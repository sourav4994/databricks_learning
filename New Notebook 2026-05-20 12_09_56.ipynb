```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, lit, when, avg, round
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, DoubleType

# Setup session (Skip this step if using Databricks)
spark = SparkSession.builder \
    .appName("NotebookDataFrameDemo") \
    .getOrCreate()

# 1. Define explicit schema 
schema = StructType([
    StructField("employee_id", IntegerType(), True),
    StructField("name", StringType(), True),
    StructField("department", StringType(), True),
    StructField("salary", IntegerType(), True),
    StructField("country", StringType(), True)
])

# 2. Create mock raw data
data = [
    (101, "Alice", "Engineering", 120000, "US"),
    (102, "Bob", "Engineering", 95000, "IN"),
    (103, "Cathy", "HR", 70000, "US"),
    (104, "David", "Finance", 110000, "UK"),
    (105, "Emma", "Engineering", 140000, "US"),
    (106, "Frank", "HR", 65000, "IN")
]

# 3. Build the DataFrame
df_raw = spark.createDataFrame(data, schema)

# In Databricks use: display(df_raw)
# In standard Jupyter notebooks use:
df_raw.show()
```
